### Section A Data Structures & Control Structures

```python
num=[5,2,4,8,9]

print(num)

num.append(10)
print("After Append",num)

num.extend([11,12,13])
print("After Extend",num)

num.insert(2,15)
print("After Insert",num)

num.remove(4)
print("After Remove",num)

removed=num.pop()

print("After Pop",num)
print("Removed Element",removed)

print("Index of 8 is",num.index(8))

print("Count of 9 is",num.count(9))

num.sort()
print("After Sort",num)

num.reverse()
print("After Reverse",num)
```

```python
students=("rahul","uday","sachin")
print("Original students",students)

students[1]="vijay" 
print("After Change",students)

```
