# Estimation And Confidence Intervals


### Data

Durability is measured in millions of characters.

There are 15 print-heads in the sample.

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
n = len(data)
print("Sample size =", n)
```

## a. 99% Confidence Interval Using Sample Standard Deviation

The population standard deviation is not known and the sample size is small. So, the t-distribution is used.

Degrees of freedom = n - 1 = 14.

```python
import math

mean = sum(data) / n
s = math.sqrt(sum((x - mean) ** 2 for x in data) / (n - 1))

print("Sample mean =", round(mean, 4))
print("Sample standard deviation =", round(s, 4))
```

For a 99% confidence interval with 14 degrees of freedom, t critical value is approximately 3.012.


```python
t_critical = 3.012
margin = t_critical * s / math.sqrt(n)
lower = mean - margin
upper = mean + margin

print("Margin of error =", round(margin, 4))
print("99% Confidence Interval =", (round(lower, 4), round(upper, 4)), "million characters")
```

### Answer for (a)

99% confidence interval = **(1.088, 1.389) million characters**.

## b. 99% Confidence Interval Using Known Population Standard Deviation

The population standard deviation is given as 0.2 million characters. Therefore, the z-distribution is used.

For a 99% confidence interval, z critical value is 2.576.



```python
sigma = 0.2
z_critical = 2.576

margin_z = z_critical * sigma / math.sqrt(n)
lower_z = mean - margin_z
upper_z = mean + margin_z

print("Population standard deviation =", sigma)
print("Margin of error =", round(margin_z, 4))
print("99% Confidence Interval =", (round(lower_z, 4), round(upper_z, 4)), "million characters")
```

### Answer for (b)

99% confidence interval = **(1.106, 1.372) million characters**.
