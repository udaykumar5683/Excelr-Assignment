# Hypothesis Testing


### Given data

Sample size = 25 restaurants

Sample mean weekly cost = Rs. 3050

Cost model: W = 1000 + 5X

Mean number of units = 600

Standard deviation of X = 25

Significance level = 0.05

### 1. Hypothesis

H0: μ ≤ 4000

H1: μ > 4000

```python
n = 25
x_bar = 3050
x_mean = 600
sigma_x = 25
alpha = 0.05

print("H0: μ <= 4000")
print("H1: μ > 4000")
```

### 2. Theoretical mean weekly cost

W = 1000 + 5X

```python
mu = 1000 + 5 * x_mean
print("Theoretical mean = Rs.", mu)
```

### 3. Test statistic

The standard deviation of weekly cost is:

σ = 5 × 25 = 125

Standard Error = σ / √n

```python
import math

sigma = 5 * sigma_x
se = sigma / math.sqrt(n)
z = (x_bar - mu) / se

print("Sigma =", sigma)
print("Standard Error =", se)
print("Z value =", z)
```

### 4. Critical value and probability

For a right-tailed test at 5% significance level, the critical value is 1.645.

```python
critical_value = 1.645

def cdf(value):
    return 0.5 * (1 + math.erf(value / math.sqrt(2)))

p_value = 1 - cdf(z)

print("Critical value =", critical_value)
print("P-value =", p_value)
```

### 5. Decision


```python
if z > critical_value:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

### 6. Conclusion

The calculated z value is -38, which is less than the critical value 1.645. Therefore, we fail to reject H0.

There is not enough evidence to support the claim that the weekly operating cost is higher than the model suggests.
