```python
import pandas as pd
patient=pd.read_csv("Patient_Data.csv")
patient
```

```python
patient.info()
```

```python
billing_columns=patient[['PatientID', 'Department', 'Doctor', 'BillAmount']]
billing_columns
```

```python
patient.drop(columns=['ReceptionistID','CheckInTime'],inplace=True)
```

```python
patient
```

```python
department_bill=patient.groupby('Department')['BillAmount'].sum()
department_bill
```

```python
patient=patient.drop_duplicates(subset='PatientID')
patient
```

```python
mean_bill=patient['BillAmount'].mean()
patient['BillAmount']=patient['BillAmount'].fillna(mean_bill)
patient
```

```python
billing=pd.read_csv("Billing_Data.csv")
billing
```

```python
merge_data=pd.merge(patient,billing,on='PatientID',how='inner')
merge_data
```

```python
new_patient=pd.DataFrame({'PatientID':[106,107],'Name':['Jhon','Smith'],'Department':['Cardiology','Neurology'],'Doctor':['Dr.Smith','Dr.Jhon'],'BillAmount':[5000,6000]})
```

```python
updated_patient=pd.concat([patient,new_patient],axis=0,ignore_index=True)
updated_patient
```

```python
new_billing_columns = pd.DataFrame({
    'InsuranceCovered': [True, False],
    'FinalAmount': [4500, 7000]
})
```

```python
final_data = pd.concat([updated_patient, new_billing_columns], axis=1)

final_data
```
