Prime Numbers

```python
num=int(input("Enter the number"))
if num<=1:
    print("The number is not prime")
else:
    for i in range(2,num):
        if num%i==0:
            print("The number is not prime")
            break
        else:
            print("The number is prime")
            break
```

Product of Random Numbers

```python
import random
num1=random.randint(1,100)
num2=random.randint(1,100)

ans=int(input(f"what is the product of {num1} x {num2}"))

corr=num1*num2
if ans==corr:
    print("Correct Answer")
else:
    print(f"Wrong Answer, the correct answer is {corr}")

```

Squares of Even/Odd Numbers

```python
for i in range(100,200):
    if i%2==0:
        print(f"The Square of {i} is {i**2}")
```

Hospital Billing System


```python
room_charges_per_day=2000
doc_con=1500
lab_test_fee=300

medicine_charges=2400
days=4
lab_test=3

room_charges=room_charges_per_day*days
lab_charges=lab_test_fee*lab_test

total_bill=room_charges+doc_con+lab_charges+medicine_charges

if total_bill>10000:
    discount=total_bill*0.10
else:
    discount=0

final_bill=total_bill-discount
print(final_bill)
```
